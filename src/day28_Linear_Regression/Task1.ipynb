```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv("D:\DS_AI_Internship\src\day28_Advanced_ML\canada_per_capita_income.csv")
df.head()
```
