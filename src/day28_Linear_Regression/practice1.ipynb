```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("D:\DS_AI_Internship\src\day28_Advanced_ML\homeprices.csv")
df
```

```python
plt.xlabel('area')
plt.ylabel('price')
plt.scatter(df.area,df.price,color='red',marker='o')
```

```python
area=df.drop('price',axis='columns')
area
```

```python
price=df.price
price
```

```python
#Equation for Linear regression y=mx+c

reg=linear_model.LinearRegression()
reg.fit(area,price)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
#To predict the new area 2500
new_data=pd.DataFrame([[3500]],columns=['area'])
prediction=reg.predict(new_data)
prediction=np.round(prediction,2)
prediction
```
