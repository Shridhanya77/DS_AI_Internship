Types of Machine Learning : 

1. Supervised Learning

Supervised Learning is a type of Machine Learning where the model is trained using labeled data, meaning each input already has a correct output or answer. The algorithm learns the relationship between inputs and outputs so it can predict results for new unseen data. It is commonly used for classification and regression problems.


2. Unsupervised Learning

Unsupervised Learning works with unlabeled data, where no correct answers are provided. The goal is to discover hidden patterns, structures, or groups within the data. The algorithm automatically finds similarities and clusters among data points.


3. Reinforcement Learning

Reinforcement Learning is a learning method where an agent learns by interacting with an environment. The model receives rewards or penalties based on its actions and gradually learns the best strategy to maximize rewards over time.
